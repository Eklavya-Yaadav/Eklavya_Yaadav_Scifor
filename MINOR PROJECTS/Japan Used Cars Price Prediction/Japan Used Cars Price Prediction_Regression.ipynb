## Now we will discuss the problem statement

**Overview**

* The dataset encompasses a range of features that provide detailed information about used cars available on the platform. This dataset serves as a valuable resource for analyzing trends, predicting car prices, and understanding the factors that influence the used car market in Japan.

**Dataset Description**

* The dataset includes the following ten features for each car:

**Independent Variables (Features)**          

Mark             -   The brand of car

Model            -   model of the vehicle

Years            -   The vehicle registration year

Mileage          -   miles traveled by vehicle

Engine_capacity  -   The measurement of the total volume of the cylinders in the engine

Transmission     -   The type of gearbox used by the car

Drive            -   wheel drive(2wd, 4wd and awd)

Hand_drive       -   Left-hand traffic (LHT) and right-hand traffic(RHT)

Fuel             -   The type of fuel used by the car( gasoline, diesel,hybrid, lpg and cng )


**Dependent Variable (Target Variable)**

Price            -   The sale price of the vehicle in the ad

```python
# Outliers Detection using Boxplot

continous_variable = data[["price","year","mileage","engine_capacity"]]

plt.figure(figsize=(20,10),facecolor='w')
plotnumber=1
for i in continous_variable:
    if plotnumber <=4:
        plt.subplot(2,3,plotnumber)
        ax=sns.boxplot(data=data[i])
        plt.xlabel(i,fontsize=20)
    plotnumber +=1
plt.tight_layout()
```

* Outliers are present we are going to remove these outliers because it will impact our model performance.

* Finding the IQR (Inter Quantile Range) to identify outliers

```python
#1st quantile 

q1 = data.select_dtypes(include = "number").quantile(0.25)
q1
```

```python
q3 = data.select_dtypes(include = "number").quantile(0.75)
q3
```

```python
# IQR Formula:
 
IQR = q3  - q1
IQR
```

# Outliers Detection Formula
Upper Cut-off ==> Q3 + (1.5 * IQR)

Lower Cut-off ==> Q1 - (1.5 * IQR)

```python
# Detecting outlier for Price variable 

print("p0   = ",data.price.min())
print("p25  = ",data.price.quantile(0.25))
print("p50  = ",data.price.median())
print("p75  = ",data.price.quantile(0.75))
print("p100 = ",data.price.max())
print("LC   = ",data.price.quantile(0.25)-1.5*IQR.price)
print("UC   = ",data.price.quantile(0.75)+1.5*IQR.price)
```

```python
# Here, we clearly see that there is a Outlier present in the Price variable at Lower Cut-off.
# Treatment of Outlier :

data.price.clip(lower = data.price.quantile(0.25)-1.5*IQR.price, inplace=True)

# here we are not dropping that observation (outlier datapoint), we are just replacing it with LC 
```

```python
# Detecting outlier for Mileage variable 

print("p0   = ",data.mileage.min())
print("p25  = ",data.mileage.quantile(0.25))
print("p50  = ",data.mileage.median())
print("p75  = ",data.mileage.quantile(0.75))
print("p100 = ",data.mileage.max())
print("LC   = ",data.mileage.quantile(0.25)-1.5*IQR.mileage)
print("UC   = ",data.mileage.quantile(0.75)+1.5*IQR.mileage)
```

```python
# Here, we clearly see that there is a Outlier present in the Mileage variable at 
# Upper Cut-off 
# Treatment of Outlier :

data.mileage.clip(upper = data.mileage.quantile(0.75) + 1.5 * IQR.mileage, inplace=True)

# here we are not dropping that observation (outlier datapoint), we are just replacing it with UC 
```

```python
# Detecting outlier for Engine Capacity variable

print("p0   = ",data.engine_capacity.min())
print("p25  = ",data.engine_capacity.quantile(0.25))
print("p50  = ",data.engine_capacity.median())
print("p75  = ",data.engine_capacity.quantile(0.75))
print("p100 = ",data.engine_capacity.max())
print("LC   = ",data.engine_capacity.quantile(0.25)-1.5*IQR.engine_capacity)
print("UC   = ",data.engine_capacity.quantile(0.75)+1.5*IQR.engine_capacity)
```

```python
# Here, we clearly see that there is a Outlier present in the Engine Capacity variable at 
# Upper Cut-off and Lower Cut-off
# Treatment of Outlier :

data.engine_capacity.clip(lower = data.engine_capacity.quantile(0.25)-1.5*IQR.engine_capacity,
                        upper=data.engine_capacity.quantile(0.75)+1.5*IQR.engine_capacity,inplace=True)

# here we are not dropping that observation (outlier datapoint), we are just replacing it with UC and LC 
```

* All the VIF are less then 5 and are very low, That means no multicollinearity is present.Now, we can go ahead with fitting our data to the model. Before that lets split our data in Train and Test.

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, classification_report
import statsmodels.api as sm
from sklearn.linear_model import Ridge
from sklearn.model_selection import RandomizedSearchCV
from sklearn.linear_model import Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
```

## Now train, test and split the data

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled,y,test_size=0.2,random_state=42)
X_train.shape, X_test.shape
```

# Model Selection

```python
import numpy as np
```

```python
def evaluate_model(true, predicted):
    mae = mean_absolute_error(true, predicted)
    mse = mean_squared_error(true, predicted)
    rmse = np.sqrt(mean_squared_error(true, predicted))
    r2_square = r2_score(true, predicted)
    return mae, rmse, r2_square
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Lasso": Lasso(),
    "Ridge": Ridge(),
    "Decision Tree": DecisionTreeRegressor(),
    "Random Forest Regressor": RandomForestRegressor()
}
model_list = []
r2_list =[]

for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train, y_train) # Train model

    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Evaluate Train and Test dataset
    model_train_mae , model_train_rmse, model_train_r2 = evaluate_model(y_train, y_train_pred)

    model_test_mae , model_test_rmse, model_test_r2 = evaluate_model(y_test, y_test_pred)


    print(list(models.keys())[i])
    model_list.append(list(models.keys())[i])

    print('Model performance for Training set')
    print("- Root Mean Squared Error: {:.4f}".format(model_train_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_train_mae))
    print("- R2 Score: {:.4f}".format(model_train_r2))

    print('----------------------------------')

    print('Model performance for Test set')
    print("- Root Mean Squared Error: {:.4f}".format(model_test_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_test_mae))
    print("- R2 Score: {:.4f}".format(model_test_r2))
    r2_list.append(model_test_r2)

    print('='*35)
    print('\n')
```

# Models Performance

```python
pd.DataFrame(list(zip(model_list, r2_list)), columns=['Model Name', 'R2_Score']).sort_values(by=["R2_Score"],ascending=False)
     
```

# Hyper Parameter Tuning

```python
rf_params = {"max_depth": [5, 8, 15, 12, 10],
             "max_features": [5, 7, 17, 8],
             "min_samples_split": [2, 8, 15, 20],
             "n_estimators": [100, 200, 500, 1000]}
```

```python
randomcv_models = [("RF", RandomForestRegressor(), rf_params)]     
```

```python
from sklearn.model_selection import RandomizedSearchCV

model_param = {}
for name, model, params in randomcv_models:
    random = RandomizedSearchCV(estimator=model,
                                   param_distributions=params,
                                   n_iter=100,
                                   cv=3,
                                   verbose=2,
                                   n_jobs=-1)
    random.fit(X_train, y_train)
    model_param[name] = random.best_params_

for model_name in model_param:
    print(f"---------------- Best Params for {model_name} -------------------")
    print(model_param[model_name])
```

## Retraining the Model with best Parameters

```python
models = {
    "Random Forest Regressor": RandomForestRegressor(**model_param['RF'], n_jobs=-1)
}

model_list = []
r2_list =[]

for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train, y_train) # Train model

    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    model_train_mae , model_train_rmse, model_train_r2 = evaluate_model(y_train, y_train_pred)

    model_test_mae , model_test_rmse, model_test_r2 = evaluate_model(y_test, y_test_pred)

    print(list(models.keys())[i])
    model_list.append(list(models.keys())[i])

    print('Model performance for Training set')
    print("- Root Mean Squared Error: {:.4f}".format(model_train_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_train_mae))
    print("- R2 Score: {:.4f}".format(model_train_r2))

    print('----------------------------------')

    print('Model performance for Test set')
    print("- Root Mean Squared Error: {:.4f}".format(model_test_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_test_mae))
    print("- R2 Score: {:.4f}".format(model_test_r2))
    r2_list.append(model_test_r2)

    print('='*35)
    print('\n')
```

```python
pd.DataFrame(list(zip(model_list, r2_list*100)), columns=['Model Name', 'R2_Score']).sort_values(by=["R2_Score"],ascending=False)
```
